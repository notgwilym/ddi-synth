```python
import random
from ddi.manifest import load_dataset, write_dataset
from ddi.run import run_training

synth_id = "20260727-094204-79b898"   # v11 synthetic set

# --- source datasets ---
v11_inst, _   = load_dataset(synth_id)          # your v11 balanced set (positives + a little NONE)
coneg_inst, _ = load_dataset(cid)               # the ~16k co-mention NONE pool

pos      = [r for r in v11_inst if r["label"] != "NONE"]
v11_none = [r for r in v11_inst if r["label"] == "NONE"]
pool_none = coneg_inst[:]                        # all coneg, all NONE

print(f"positives: {len(pos)}   v11 NONE: {len(v11_none)}   coneg NONE pool: {len(pool_none)}")

# cfg: reuse your training config; force a fixed seed so the curve isolates ratio, not seed noise
BASE = {"model_name": "microsoft/BiomedNLP-BiomedBERT-base-uncased-abstract-fulltext",
        "epochs": 3, "lr": 2e-5, "batch_size": 32, "max_length": 256,
        "neg_ratio": None, "seed": 0, "dataset": "synthetic"}

def none_needed(n_pos, target):
    # target = NONE / (NONE + pos)  ->  NONE = target/(1-target) * pos
    return int(round(target / (1 - target) * n_pos))

TARGETS = [0.30, 0.50, 0.70, 0.85]
rng = random.Random(0)
results = []

for t in TARGETS:
    need = none_needed(len(pos), t)
    have = len(v11_none) + len(pool_none)
    if need > have:
        print(f"  ratio {t}: need {need} NONE but only have {have} -- capping (max ratio {have/(have+len(pos)):.2f})")
        need = have
    # take all v11 NONE first, top up from the coneg pool
    extra = max(0, need - len(v11_none))
    none_slice = v11_none + rng.sample(pool_none, min(extra, len(pool_none)))
    combined = pos + none_slice
    rng.shuffle(combined)

    actual = len(none_slice) / len(combined)
    ds_id = write_dataset(
        combined, provenance="synthetic:v11+coneg-mix",
        generator={"note": f"mixing curve, target NONE {t}", "prompt_sha": prompt_fingerprint()},
        seed=0, notes=f"mix target={t} actual={actual:.3f}")
    run_id, m = run_training(ds_id, dev_id, cfg=BASE, notes=f"mix NONE={t}")
    results.append((t, actual, m["micro_f1_pos"], m["micro_p_pos"], m["micro_r_pos"]))
    print(f"  NONE {t:.2f} (actual {actual:.2f})  F1={m['micro_f1_pos']:.3f}  "
          f"P={m['micro_p_pos']:.3f}  R={m['micro_r_pos']:.3f}")

print("\n=== mixing curve ===")
print(f"{'target':>7} {'actual':>7} {'F1':>6} {'P':>6} {'R':>6}")
for t, a, f, p, r in results:
    print(f"{t:>7.2f} {a:>7.3f} {f:>6.3f} {p:>6.3f} {r:>6.3f}")
```

```python
def fpr(rows, name):
    print(f"\n{name}")
    print(f"{'bucket':<8} {'pairs':>6} {'gold+':>6} {'pred+':>6} {'FP':>6} {'neg':>6} {'FPR':>7} {'pos%':>7}")
    for r in rows:
        correct = r["P"] * r["predicted"]
        fp = r["predicted"] - correct
        neg = r["n_pairs"] - r["support"]
        print(f"{r['bucket']:<8} {r['n_pairs']:>6} {r['support']:>6} {r['predicted']:>6} "
              f"{fp:>6.0f} {neg:>6} {fp/neg if neg else 0:>7.3f} "
              f"{r['support']/r['n_pairs']:>7.3f}")

fpr(train, "human"); fpr(sy_inst, "synthetic")
```
